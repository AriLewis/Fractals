Compute the Mandelbrot set, $f_c(z)=z^2+c$, following this tutorial: https://tomroelandts.com/articles/how-to-compute-the-mandelbrot-set-using-numpy-array-operations

$c$ is taken at every complex number. We will let it be the coordinates of the plot. Then iterate on $f_c$, starting with $z=0$.

```python
m = 1024
n = 680

x = np.linspace(-2, 1, num=m).reshape((1, m))
y = np.linspace(-1, 1, num=n).reshape((n, 1))
C = np.tile(x, (n, 1)) + 1j * np.tile(y, (1, m))
```

```python
fig = plt.figure()
fig.set_size_inches(m / 100, n / 100)
ax = fig.add_axes([0, 0, 1, 1], frameon=False, aspect=1)
ax.set_xticks([])
ax.set_yticks([])
plt.imshow(np.abs(C))
plt.show()
```

```python
Z = np.zeros((n, m), dtype=complex)
M = np.full((n, m), True, dtype=bool)
N = np.zeros((n, m))
for i in range(100):
    Z[M] = Z[M]**2 + C[M]
    M[np.abs(Z) > 2] = False
    N[M] = i
```

```python
fig = plt.figure()
fig.set_size_inches(m / 100, n / 100)
ax = fig.add_axes([0, 0, 1, 1], frameon=False, aspect=1)
ax.set_xticks([])
ax.set_yticks([])
plt.imshow(N-1)
plt.show()
```

Compute the Julia set, also $f_c(z)=z^2+C$ https://tomroelandts.com/articles/how-to-compute-colorful-fractals-using-numpy-and-matplotlib

```python
from __future__ import division

# m = 480
# n = 320

s = 1000 # Scale
x = np.linspace(-m / s, m/ s, num=m).reshape((1, m))
y = np.linspace(-n / s, n /s, num=n).reshape((n, 1))
Z = np.tile(x, (n, 1)) +1j * np.tile(y, (1,m))
```

```python
C = np.full((n, m), -0.4 + 0.6j)
M = np.full((n, m), True, dtype=bool)
N = np.zeros((n,m))
for i in range(256):
    Z[M] = Z[M] **2 + C[M]
    M[np.abs(Z)  >  2] = False
    N[M] = i
```

```python
fig = plt.figure()
fig.set_size_inches(m / 100, n / 100)
ax = fig.add_axes([0, 0, 1, 1], frameon=False, aspect=1)
ax.set_xticks([])
ax.set_yticks([])
plt.imshow(N, cmap='hot')
plt.savefig('julia-plt.png')
plt.show()
```

```python
s = 1000  # Scale.
x = np.linspace(-m / s, m / s, num=m).reshape((1, m))
y = np.linspace(-n / s, n / s, num=n).reshape((n, 1))
Z = np.tile(x, (n, 1)) + 1j * np.tile(y, (1, m))
 
C = np.full((n, m), -0.45 + 0.6j)
M = np.full((n, m), True, dtype=bool)
N = np.zeros((n, m))
for i in range(256):
    Z[M] = Z[M] * Z[M] + C[M]
    M[np.abs(Z) > 2] = False
    N[M] = i

# Save with Matplotlib using a colormap.
fig = plt.figure()
fig.set_size_inches(m / 100, n / 100)
ax = fig.add_axes([0, 0, 1, 1], frameon=False, aspect=1)
ax.set_xticks([])
ax.set_yticks([])
plt.imshow(np.flipud(N), cmap='hot')
plt.show()
```
